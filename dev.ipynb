```python
from dotenv import load_dotenv
import os
from openai import OpenAI
```

```python
load_dotenv()
client = OpenAI()
```

```python
import base64
import requests
```

```python
# Function to encode the image
def encode_image(image_path):
  with open(image_path, "rb") as image_file:
    return base64.b64encode(image_file.read()).decode('utf-8')
```

```python
def save_to_markdown(response_text, filename):
    with open(filename, "w") as f:
        f.write(response_text)
```

```python
api_key = os.getenv("OPENAI_API_KEY")
```

```python
query = """
Objective: Parse the cooking recipe from the image.

Context: The recipe includes separate lists for marinating and cooking ingredients.

Instructions:

Split the ingredients into two categories: Marinate Ingredients and Cooking Ingredients.
Format the recipe in markdown as follows:
Header: Recipe title
Ingredients: List the ingredients for marinating and cooking, using * for each ingredient.
Preparation: Provide the steps in point form.
How to cook: Provide the cooking steps in point form.
Recipe: Parse and format the content of the recipe image.
"""
```

```python
query = """
Objective: Parse the cooking recipe from the image.

Instructions:

Format the recipe in markdown as follows:
Header: Recipe title
Ingredients: List the ingredients needed
Instructions: Provide the preparation and cooking steps in point form.
"""
```

```python
image_path = "./img/xiao-bai-cai.jpeg"
base64_image = encode_image(image_path)

headers = {
  "Content-Type": "application/json",
  "Authorization": f"Bearer {api_key}"
}

payload = {
  "model": "gpt-4o-mini",
  "messages": [
    {
      "role": "user",
      "content": [
        {
          "type": "text",
          "text": f"{query}"
        },
        {
          "type": "image_url",
          "image_url": {
            "url": f"data:image/jpeg;base64,{base64_image}"
          }
        }
      ]
    }
  ],
  "max_tokens": 300
}

response = requests.post("https://api.openai.com/v1/chat/completions", headers=headers, json=payload)

save_to_markdown(response.json()["choices"][0]["message"]["content"], "./output/stir-fried-pork.md")
```

```python
print(response.json()['choices'][0]['message']['content'])
```
